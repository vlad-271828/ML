# Методы опорных векторов
Метод опорных векторов (Support Vector Machine - SVM) - это очень мощная и универсальная модель машинного обучения, способная выполнять линейную или нелинейную классификацию, регрессию и даже выявление выбросов.

Методы SVM особенно хорошо подходят для классфикации сложных, но небольших или средних наборов данных.

# Линейная классификация SVM

_Фундаментальная идея SVM_: классификатор SVM устнавливает самую широкую, какую только возможно, полосу между классами - классификация с широким зазором.

Методы SVM чувствительны к масштабам признаков!
